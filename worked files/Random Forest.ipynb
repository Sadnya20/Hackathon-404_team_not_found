```python
import pandas as pd

# Load the uploaded aviation dataset
file_path = "Aviation_KPIs_POWERBI.csv"
df = pd.read_csv(file_path)

# Display basic info and first few rows
df.info(), df.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score

# Select features and target
target = 'Profit (USD)'
features = df.select_dtypes(include=['float64', 'int64']).drop(columns=[target])

X = features
y = df[target]

# Scale features using MinMaxScaler
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# Train a RandomForestRegressor
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# Predict and evaluate
y_pred = model.predict(X_test)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

mse, r2
```

```python
# Sample 10,000 rows for faster training
df_sampled = df.sample(n=10000, random_state=42)

# Prepare features and target again
X_sampled = df_sampled.select_dtypes(include=['float64', 'int64']).drop(columns=[target])
y_sampled = df_sampled[target]

# Scale features
X_sampled_scaled = scaler.fit_transform(X_sampled)

# Split into train and test sets
X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(X_sampled_scaled, y_sampled, test_size=0.2, random_state=42)

# Train the RandomForestRegressor on sampled data
model_sampled = RandomForestRegressor(n_estimators=100, random_state=42)
model_sampled.fit(X_train_s, y_train_s)

# Predict and evaluate
y_pred_s = model_sampled.predict(X_test_s)
mse_s = mean_squared_error(y_test_s, y_pred_s)
r2_s = r2_score(y_test_s, y_pred_s)

mse_s, r2_s
```
